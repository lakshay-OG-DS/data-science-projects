* **About the data** - Overview
This dataset combines character stats/information (scraped from the Genshin Impact Wiki) with aggregate gacha pull and constellation data (scraped from paimon.moe) to give a full picture of each character's in-game stats alongside community pull trends.

Contents:
Character stats and information
Total pull counts per character
Constellation distribution per character
Character banners and related temporal information
Derived statistics: average copies per player, duplicate rate, C6 rate calculated from the aggregated pull/constellation data

```python
# Cleaning and fixing the data
# Converting the data type of months and copies
df['months_since_release'] = df['months_since_release'].astype(int)
df['days_since_release'] = df['days_since_release'].astype(int)
df['avg_copies_per_player'] = df['avg_copies_per_player'].astype(int)
df['year'] = pd.to_datetime(df['release_date']).dt.year
```

# SQL

```python
conn = sqlite3.connect('genshin.db')
df.to_sql('genshin_data',conn,index = False , if_exists = 'replace')
```

```python
# Format - each query will annswer one question - 

# QUE 1 - Average lvl 90 ATK AND DEF
query1 = 'SELECT AVG("lvl_90_ATK") as avg_atk , AVG("lvl_90_DEF") as avg_def FROM genshin_data'
print('AVERAGE 90 LVL ATK POINTS : \n',pd.read_sql_query(query1,conn))


# QUE 2 - Element with most 5 stars
query2 = 'SELECT "element" , COUNT(*) as "5_STAR_COUNT" FROM genshin_data GROUP BY "element" HAVING "rarity"> 4 ORDER BY "5_STAR_COUNT" DESC'
print('Element with the highest number of 5 stars :\n',pd.read_sql_query(query2,conn))

# QUE 3 - Characters with Lowest and highest number of pulls
query3 = 'SELECT "name" , MIN("pulled_count") FROM genshin_data GROUP BY name ORDER BY "pulled_count" ASC LIMIT 3 '
query4 = 'SELECT "name" , MAX("pulled_count") FROM genshin_data GROUP BY name ORDER BY "pulled_count" DESC LIMIT 3 '
print('Character  pull count : \n',pd.read_sql_query(query4,conn))
print('Character  pull count : \n',pd.read_sql_query(query3,conn))

# QUE 4 -  Characters with Lowest and highest number of banners
query3 = 'SELECT "name" , MIN("num_banners") FROM genshin_data GROUP BY name ORDER BY "num_banners" ASC LIMIT 3 '
query4 = 'SELECT "name" , MAX("num_banners") FROM genshin_data GROUP BY name ORDER BY "num_banners" DESC LIMIT 3 '
print('Character  pull count : \n',pd.read_sql_query(query4,conn))
print('Character  pull count : \n',pd.read_sql_query(query3,conn))

# QUE 5 - Characters with Lowest and Highest C6 Rates

# Query for Lowest C6 Rate
query5_min = 'SELECT "name", MIN("c6_rate") AS "c6_rate" FROM genshin_data GROUP BY "name" ORDER BY "c6_rate" ASC LIMIT 3'

# Query for Highest C6 Rate
query5_max = 'SELECT "name", MAX("c6_rate") AS "c6_rate" FROM genshin_data GROUP BY "name" ORDER BY "c6_rate" DESC LIMIT 3'

print('Characters with Lowest C6 Rate:\n', pd.read_sql_query(query5_min, conn))
print('\nCharacters with Highest C6 Rate:\n', pd.read_sql_query(query5_max, conn))
```

# ML MODEL FOR CHARACTER RARITY PREDICTION

```python
# Machine learning model
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
```

```python
# Convert text/categorical columns into numeric dummy columns
x = df[['element','region','roles','lvl_90_ATK','lvl_90_DEF','lvl_90_HP','num_banners','is_archon']]
y = df['rarity']

x = pd.get_dummies(x, columns=['element', 'region', 'roles'])

x_train,x_test,y_train,y_test = train_test_split(x,
                                                y,
                                                 test_size = 0.2,
                                                random_state = 42,
                                                )

model = DecisionTreeClassifier(min_samples_split = 5,
                               min_samples_leaf = 4,
                               max_depth = 12,
                               random_state = 42
                              )
model.fit(x_train,y_train)
predictions = model.predict(x_test)
print('Model prediction :',accuracy_score(y_test,predictions))

scores = cross_val_score(model, x, y, cv=5)
print('Cross-Validation Accuracy Scores:', scores)
print('Average Accuracy:', scores.mean())
```
